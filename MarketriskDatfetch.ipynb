```python
import yfinance as yf
import pandas as pd
import os

tickers = ["^GSPC", "^VIX"]
data = yf.download(tickers=tickers, start="2019-01-01", end="2025-10-26", interval="1d")

save_path = r"C:\Users\Venu Shambhu\Desktop\Capstone Project\Raw Data\Marketrisk\market_data_^GSPC_^VIX.csv"
os.makedirs(os.path.dirname(save_path), exist_ok=True)
data.to_csv(save_path)

print(f"✅ Data saved successfully at: {save_path}")
```
